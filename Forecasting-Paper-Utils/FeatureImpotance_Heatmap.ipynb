```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy.interpolate import make_interp_spline
import random

# Read the CSV file
df = pd.read_csv('data.csv')

# Set up the plot
plt.figure(figsize=(12, 8))

# Colors for each line
colors = ['red', 'blue', 'green', 'purple']

# Loop through each column
for i, column in enumerate(df.columns):
    # Get the data
    y = df[column].dropna().values
    x = np.arange(len(y))

    # Create a smooth line
    # x_smooth = np.linspace(x.min(), x.max(), 300)
    # spl = make_interp_spline(x, y, k=3)
    # y_smooth = spl(x_smooth)
    x_smooth = np.arange(len(y))
    y_smooth = y

    # Plot the smooth line
    plt.plot(x_smooth, y_smooth, color=colors[i], label=column)

    # Randomly mark 10 points
    column_length = len(x)
    random_indices = random.sample(range(column_length), 10)
    plt.scatter(x[random_indices], y[random_indices], color=colors[i], s=50, zorder=5)

# Customize the plot
plt.title('Smoothed Line Curves with 10 Random Points Marked')
plt.xlabel('Index')
plt.ylabel('Value')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)

# Show the plot
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Read the CSV file
df = pd.read_csv('data.csv', header=None)

# Create a figure and axis
fig, ax = plt.subplots(figsize=(10, 6))

# Define line styles and colors for each curve
line_styles = [':', '-.', '--', ':']
colors = ['red', 'blue', 'green', 'purple']

# Plot each column as a dotted line curve
for i in range(4):
    # Get the non-null values from the column
    values = df[i].dropna().values
    x = np.arange(len(values))

    # Plot the curve

    ax.plot(x, values, linestyle=line_styles[i], color=colors[i])

# Customize the plot
ax.set_xlabel('Index')
ax.set_ylabel('Prediction Value')
ax.set_title('Prediction Curves')
ax.legend()
ax.grid(True, linestyle='--', alpha=0.7)

# Show the plot
plt.tight_layout()
plt.show()
```
