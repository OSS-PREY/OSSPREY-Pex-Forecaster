```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the datasets
# Load the datasets
file1 = 'clean-apache-network-data-2-2.csv'
file2 = 'clean-eclipse-network-data-3-3.csv'
file3 = 'clean-osgeo-network-data-2-2.csv'
file4 = 'clean-github-network-data-4-5.csv'


df1 = pd.read_csv(file1)
df2 = pd.read_csv(file2)
df3 = pd.read_csv(file3)
df4 = pd.read_csv(file3)

# Ensure the month column is in datetime format
df1['month'] = pd.to_datetime(df1['month'])
df2['month'] = pd.to_datetime(df2['month'])
df3['month'] = pd.to_datetime(df3['month'])
df4['month'] = pd.to_datetime(df4['month'])



# Calculate the project lengths for each dataset
project_lengths1 = df1.groupby('proj_name')['month'].nunique()
project_lengths2 = df2.groupby('proj_name')['month'].nunique()
project_lengths3 = df3.groupby('proj_name')['month'].nunique()
project_lengths4 = df4.groupby('proj_name')['month'].nunique()

project_lengths1 = project_lengths1[project_lengths1 <= 100]
project_lengths2 = project_lengths2[project_lengths2 <= 100]
project_lengths3 = project_lengths3[project_lengths3 <= 100]
project_lengths4 = project_lengths4[project_lengths4 <= 100]


################# NEW EDITS STARTS #####################

# df = pd.read_csv("eclipse_stats_temp.csv", parse_dates=["START DATE", "END DATE"])

# # Calculate lifetime duration in months
# df["LIFETIME (months)"] = (df["END DATE"] - df["START DATE"]).dt.days // 30

# # Store as a Pandas Series
# project_lengths2 = df.set_index("PROJECT")["LIFETIME (months)"]

# # Display the series
# print(project_lengths2)




# df = pd.read_csv("OSGEo_stats_temp.csv", parse_dates=["START DATE", "END DATE"])

# # Calculate lifetime duration in months
# df["LIFETIME (months)"] = (df["END DATE"] - df["START DATE"]).dt.days // 30

# # Store as a Pandas Series
# project_lengths3 = df.set_index("PROJECT")["LIFETIME (months)"]

# # Display the series
# print(project_lengths3)

################# NEW EDITS ENDS #####################


# Plot the distributions of project lengths as KDE plots
plt.figure(figsize=(8, 4))

sns.kdeplot(project_lengths1, label='Apache', bw_adjust=0.5, fill=True, common_norm=False)
sns.kdeplot(project_lengths2, label='Eclipse', bw_adjust=2, fill=True, common_norm=False)
sns.kdeplot(project_lengths3, label='OSGEO', bw_adjust=0.3, fill=True, common_norm=False)
sns.kdeplot(project_lengths4, label='GitHub', bw_adjust=0.5, fill=True, common_norm=False)

plt.ylim(0, 0.050)  # Only show density up to 0.5

# plt.title('Distribution of Project Lengths (in months)')
plt.xlabel('Project Length (months)', fontsize=8)
plt.ylabel('Density', fontsize=8)
plt.legend(loc='upper right', fontsize = 8)
plt.grid(axis='y', linestyle='--', alpha=0.9)

# fig, ax = plt.subplots()

# Change font size of tick labels
plt.tick_params(axis='both', which='major', labelsize=6)
plt.savefig('project_len_dist.png', dpi=400, bbox_inches='tight')
plt.show()
```
